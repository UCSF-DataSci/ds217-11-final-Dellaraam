## Setup

````python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load feature-engineered data from Q4
df = pd.read_csv('output/q4_features.csv', parse_dates=['Measurement Timestamp'], index_col='Measurement Timestamp')
# Or if you saved without index:
# df = pd.read_csv('output/q4_features.csv')
# df['Measurement Timestamp'] = pd.to_datetime(df['Measurement Timestamp'])
# df = df.set_index('Measurement Timestamp')
print(f"Loaded {len(df):,} records with features")
````

````python
matrix = pd.DataFrame({
    'Maximum Wind Speed': df['Maximum Wind Speed'], 
    'Air Temperature' : df['Air Temperature'],
    'Humidity' : df['Humidity'],
    'Total Rain' : df['Total Rain']
})
matrix_corr = matrix.corr()
display(matrix_corr)
matrix_corr.to_csv('output/q5_correlations.csv')
````

````python
fig, axes = plt.subplots(2, 2, figsize=(25, 15))
start_date = df.index.min()
end_date = df.index.max()

monthly_rain = df['Total Rain'].resample('M').mean()
monthly_air = df['Air Temperature'].resample('M').mean()
monthly_hum = df['Humidity'].resample('M').mean()
monthly_max_wind = df['Maximum Wind Speed'].resample('M').mean()
axes[0][0].plot(monthly_rain)
axes[0][0].plot(monthly_air)
axes[0][0].plot(monthly_hum)
axes[0][0].plot(monthly_max_wind)
axes[0][0].set_title('Total Monthly Rain')
axes[0][0].set_xlabel('Date')
axes[0][0].set_ylabel('Rain')
axes[0][0].legend(["monthly_rain", "monthly_air", "monthly_hum", "monthly_max_wind"])
axes[0][0].grid(True)

monthly_rain = df['Total Rain'].resample('M').mean()
daily_rain = df['Total Rain'].resample('D').mean()
yearly_rain = df['Total Rain'].resample('Y').mean()
axes[1][0].plot(monthly_rain)
axes[1][0].plot(daily_rain)
axes[1][0].plot(yearly_rain)
axes[1][0].set_title('Seasonal Rain')
axes[1][0].set_xlabel('Date')
axes[1][0].set_ylabel('Rain')
axes[1][0].legend(["monthly", "daily", "yearly"])
axes[1][0].grid(True)

axes[0][1] = sns.heatmap(data=matrix_corr)
axes[0][1].set_title('correlation numbers')

plt.tight_layout()
plt.savefig('output/q5_patterns.png', dpi=150, bbox_inches='tight')
plt.show()
````
